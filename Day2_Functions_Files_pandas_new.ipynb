## Part 1: Functions

On Day 1 you calculated GC content several times by writing the same lines again. A **function** lets you write that code once, give it a name, and reuse it.

You have already used built-in functions such as `print()`, `len()` and `max()`. Now you will write your own.

```python
def function_name(parameter):
    # code that uses the parameter
    return result
```

- `def` starts the function.
- **Parameters** are the inputs, listed in the brackets.
- `return` sends the result back to the code that called the function.
- The code inside must be indented.

````python
def gc_content(seq):
    gc = seq.count("G") + seq.count("C")
    return gc / len(seq) * 100
````

Running the cell above only **defines** the function. Nothing is printed yet. Now **call** it:

````python
print(gc_content("ATGGCGTACGCT"))
print(gc_content("ATATATAT"))

result = gc_content("GGGCCC")
print("GC content:", result,"%")
````

### Several parameters

Separate parameters with commas. Give them clear names.

````python
def mass_g(moles, mol_weight):     #calculate the mass in grams
    return moles * mol_weight

mass_g(0.0005, 194.19)     # 0.5 mmol caffeine
````

### Default values

A parameter can have a **default**. If you do not pass a value, the default is used.

````python
def passes_mw(mw, cutoff=500):
    return mw <= cutoff

print(passes_mw(450))            # uses cutoff=500
print(passes_mw(450, cutoff=400))
````

### Docstrings

A **docstring** is a short description in triple quotes on the first line of a function.

````python
def mean_value(values):
    """Return the average of a list of numbers."""
    return sum(values) / len(values)

print(mean_value([180.16, 194.19, 206.28]))
````

### Functions inside loops

Functions and loops work well together: write the logic once, then apply it to every item.

````python
sequences = ["ATGCGTACGT", "ATATATATAT", "GGGCCCATGC"]

for seq in sequences:
    print(seq, gc_content(seq))
````

## Part 2: Using modules

A **module** is a file of ready-made functions that someone else wrote. Python comes with many modules, and you load one with `import`.

| Module | Useful for |
|---|---|
| `math` | logarithms, square roots, constants |
| `statistics` | mean, median, standard deviation |
| `random` | random numbers and random choices |

````python
import math

print(math.log10(1000))
print(math.sqrt(16))
print(math.pi)
````

You can also import only the functions you need with `from ... import ...`.

````python
from statistics import mean, median, stdev

mw_values = [180.16, 194.19, 206.28, 151.16, 162.24]
print("Mean:", round(mean(mw_values), 2))
print("Median:", median(mw_values))
print("Std dev:", round(stdev(mw_values), 2))
````

### Downloading data with `requests`

Many databases let you download data with a simple web address (URL). The `requests` library fetches the page for you, just like a browser.

**UniProt** is the main database of protein sequences. Every protein has an **accession** code. For example, `P00533` is human **EGFR**, a receptor targeted by several cancer drugs.

````python
import requests

url = "https://rest.uniprot.org/uniprotkb/P00533.fasta"
response = requests.get(url)

print("Status code:", response.status_code)    # 200 means it worked
print(response.text[:300])                      # first 300 characters
````

`response.text` is the downloaded data as one long string. It is in **FASTA** format: a header line starting with `>`, then the sequence split over several lines.

If the status code is not 200, check the URL and run the cell again.

### Writing a file

To keep the data, write it to a file. `with open(...)` opens the file and closes it automatically at the end of the block. `"w"` means **write** (this replaces anything already in the file).

````python
with open("egfr.fasta", "w") as f:
    f.write(response.text)
````

Check the folder panel on the left (**drive > MyDrive > BT_python_workshop**). You should see `egfr.fasta`.

### Reading a whole file

`"r"` means **read**. `.read()` gives the whole file as one string.

````python
with open("egfr.fasta", "r") as f:
    text = f.read()

print(text[:200])
````

### Downloading several proteins

Now combine a **loop**, a **download** and **writing a file**. The loop fetches each protein and adds it to one file. `"a"` means **append**: add to the end of the file instead of replacing it.

````python
proteins = {
    "P00533": "EGFR, cancer drug target",
    "P04637": "p53, tumour suppressor",
    "P68871": "hemoglobin beta",
    "P01308": "insulin",
    "P0DTC2": "SARS-CoV-2 spike protein",
}

open("proteins.fasta", "w").close()     # start with an empty file

for accession, description in proteins.items():
    url = f"https://rest.uniprot.org/uniprotkb/{accession}.fasta"
    response = requests.get(url)
    with open("proteins.fasta", "a") as f:
        f.write(response.text)
    print("Downloaded", accession, description)
````

### Selecting columns

One column gives a Series. A list of columns gives a smaller DataFrame.

````python
df["mw"]
````

````python
df[["name", "mw", "logp"]]
````

### Selecting rows

- `df.iloc[0]` gets a row by its **position**.
- `df[condition]` keeps only the rows where the condition is `True`. This is called **filtering**.

````python
df.iloc[0]
````

````python
df[df["mw"] < 200]
````

To combine conditions, use `&` (and) or `|` (or), and put each condition in **brackets**.

````python
df[(df["mw"] < 300) & (df["logp"] > 2)]
````

````python
df[df["drug_class"] == "NSAID"]
````

### Sorting

````python
df.sort_values("mw", ascending=False)[["name", "mw"]]
````

### Adding new columns

You can create a column from a calculation on other columns. The calculation runs on every row at once, with no loop needed.

````python
df["mw_kDa"] = df["mw"] / 1000
df["polar_atoms"] = df["hbd"] + df["hba"]
df[["name", "mw", "mw_kDa", "polar_atoms"]].head()
````

Use `.apply()` to run your **own function** on every value in a column.

````python
def size_label(mw):
    if mw < 200:
        return "small"
    elif mw < 400:
        return "medium"
    else:
        return "large"

df["size"] = df["mw"].apply(size_label)
df[["name", "mw", "size"]]
````

### Summarizing data

````python
print("Average MW:", round(df["mw"].mean(), 2))
print("Highest logP:", df["logp"].max())
df["drug_class"].value_counts()     # how many rows of each class
````

`groupby()` splits the table into groups and summarizes each one.

````python
df.groupby("size")["mw"].mean()
````

### Saving a DataFrame

`index=False` stops pandas from writing the row numbers as an extra column.

````python
small = df[df["size"] == "small"]
small.to_csv("small_compounds.csv", index=False)
pd.read_csv("small_compounds.csv")
````

## Part 5: Clinical data with pandas

pandas works the same way on clinical data as on chemical data. Here we use the **Wisconsin breast cancer dataset**, a public, de-identified dataset that comes with the scikit-learn library, so nothing needs to be downloaded.

- **569 patients**, each with a breast mass sampled by fine needle aspiration (FNA)
- **30 measurements** of the cell nuclei, calculated from digital images of the sample (size, shape, texture)
- A **diagnosis**: malignant or benign

**A note on privacy:** only use public, de-identified datasets like this one. Colab runs on Google's servers, so never upload real patient data from a hospital, clinic or research project to Colab.

````python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer(as_frame=True)
cancer = data.frame

print(cancer.shape)
cancer.head()
````

The `target` column holds the diagnosis as a number: **0 = malignant**, **1 = benign**. Numbers are hard to read, so we make a new column with words using `.map()`, which replaces each value using a dictionary.

````python
cancer["diagnosis"] = cancer["target"].map({0: "malignant", 1: "benign"})
cancer[["mean radius", "mean area", "mean concavity", "target", "diagnosis"]].head()
````

Column names with spaces need the bracket form: `cancer["mean radius"]`.

### Comparing two groups

`groupby()` makes it easy to compare patients with malignant and benign masses.

````python
cancer.groupby("diagnosis")[["mean radius", "mean area", "mean concavity"]].mean().round(2)
````
