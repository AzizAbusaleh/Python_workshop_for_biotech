## Part 1: Variables and data types

A **variable** is a name that stores a value. You create one with `=`.

```python
name = value
```

Variable names can use letters, numbers and `_`, but cannot start with a number. Use clear names like `mol_weight`, not `x`.

````python
drug_name = "aspirin"     # text (string)
mol_weight = 180.16       # decimal number (float), in g/mol
heavy_atoms = 13          # whole number (integer)
is_approved = True        # True or False (boolean)

print(drug_name, mol_weight, heavy_atoms, is_approved)
````

You can also assign multiple variables in a single line of code by separating the names and values with commas:

````python
drug_name, mol_weight = "ibuprofen", 206.28
print(drug_name)
print(mol_weight)
````

Every value has a **type**. Use `type()` to check it.

| Type | Example | Meaning |
|---|---|---|
| `int` | `13` | whole number |
| `float` | `180.16` | decimal number |
| `str` | `"aspirin"` | text, always inside quotes |
| `bool` | `True` | True or False |

````python
print(type(drug_name))
print(type(mol_weight))
print(type(heavy_atoms))
print(type(is_approved))
````

### Arithmetic

| Operator | Meaning | Example |
|---|---|---|
| `+` `-` | add, subtract | `5 - 2` |
| `*` `/` | multiply, divide | `10 / 4` gives `2.5` |
| `**` | power | `10 ** 3` gives `1000` |
| `//` | whole-number division | `10 // 4` gives `2` |
| `%` | remainder | `10 % 4` gives `2` |

A common lab calculation: **mass = moles x molecular weight**.

````python
moles = 0.002                  # mol
mass_g = moles * mol_weight    # g
print("Mass needed (g):", mass_g)

mass_mg = mass_g * 1000
print("Mass needed (mg):", mass_mg)
````

Use `round(value, digits)` to keep a sensible number of decimal places.

````python
print(round(mass_mg, 1))
````

You can update a variable by assigning a new value. `x += 1` is a short way to write `x = x + 1`.

````python
count = 0
count += 1
count += 1
print(count)
````

## Part 2: Strings

A **string** is text inside quotes. DNA sequences, protein sequences and SMILES are all strings, which makes strings very useful in biology and chemistry.

````python
dna = "ATGGCGTACGCTTAGCCGATATAA"
smiles = "CC(=O)Oc1ccccc1C(=O)O"    # aspirin

print(dna)
print("Length of the DNA sequence:", len(dna))
````

### Indexing and slicing

Each character has a position called an **index**. Python counts from **0**.

```
 A  T  G  G  C  G ...
 0  1  2  3  4  5 ...
```

- `dna[0]` gives the first character.
- `dna[-1]` gives the last character.
- `dna[0:3]` gives characters 0, 1 and 2. The end position is **not included**.

````python
print("First base:", dna[0])
print("Last base:", dna[-1])
print("First codon:", dna[0:3])
print("Last codon:", dna[-3:])
````

### Useful string methods

A **method** is a function attached to a value. You call it with a dot: `value.method()`.

| Method | What it does |
|---|---|
| `s.lower()`, `s.upper()` | change case |
| `s.count("G")` | count how often "G" appears |
| `s.replace("T", "U")` | replace text |
| `s.find("TAG")` | position of the first match (-1 if not found) |
| `s.startswith("ATG")` | True or False |

````python
print(dna.lower())
print("Number of G:", dna.count("G"))
print("Starts with ATG?", dna.startswith("ATG"))
print("Position of TAG:", dna.find("TAG"))
````

Strings cannot be changed in place. Methods like `replace()` give you a **new** string, so save it in a variable if you want to keep it.

````python
messy = "  atggcg  "
clean = messy.strip().upper()     # remove spaces, then make upper case
print(clean)
````

### Joining strings and f-strings

Use `+` to join strings. Use an **f-string** to put variables inside text. Write `f` before the quotes and put variables in `{ }`.

````python
gene = "BRCA1"
organism = "human"
print(gene + " is a " + organism + " gene")

length = len(dna)
print(f"The sequence has {length} bases")

gc = (dna.count("G") + dna.count("C")) / len(dna)
print(f"GC fraction: {gc:.2f}")      # :.2f means 2 decimal places
````

## Part 3: Lists

A **list** holds many values in order. Write it with square brackets `[ ]` and separate items with commas.

````python
drugs = ["aspirin", "caffeine", "ibuprofen", "paracetamol"]
mol_weights = [180.16, 194.19, 206.28, 151.16]    # g/mol, same order as drugs

print(drugs)
print("Number of drugs:", len(drugs))
````

Lists use the same indexing and slicing as strings.

````python
print("First drug:", drugs[0])
print("Last drug:", drugs[-1])
print("First two drugs:", drugs[0:2])
````

### Changing a list

Unlike strings, lists **can** be changed.

| Code | What it does |
|---|---|
| `lst.append(x)` | add `x` to the end |
| `lst.insert(i, x)` | add `x` at position `i` |
| `lst.remove(x)` | remove the first `x` |
| `lst[i] = x` | replace the item at position `i` |
| `lst.sort()` | sort the list in place |

````python
screen_hits = ["cpd_01", "cpd_07"]
screen_hits.append("cpd_12")
print(screen_hits)

screen_hits.remove("cpd_07")
print(screen_hits)

screen_hits[0] = "cpd_02"
print(screen_hits)
````

### Built-in functions for lists of numbers

````python
print("Lowest MW:", min(mol_weights))
print("Highest MW:", max(mol_weights))
print("Total:", sum(mol_weights))
print("Average MW:", round(sum(mol_weights) / len(mol_weights), 2))
print("Sorted:", sorted(mol_weights))
````

Use `in` to check if a value is in a list. The answer is `True` or `False`.

````python
print("caffeine" in drugs)
print("nicotine" in drugs)
````

A string can be turned into a list of characters with `list()`. This is handy for sequences.

````python
peptide = "MKTAYIAK"
residues = list(peptide)
print(residues)
````

## Part 4: Dictionaries

In Part 3 we kept names and weights in two separate lists and had to keep them in the same order. A **dictionary** is a cleaner way. It stores **key: value** pairs, so you look things up by name instead of by position.

Write it with curly brackets `{ }`.

````python
drug_mw = {
    "aspirin": 180.16,
    "caffeine": 194.19,
    "ibuprofen": 206.28,
    "paracetamol": 151.16,
}

print(drug_mw["caffeine"])
````

### Adding, changing and checking

````python
drug_mw["nicotine"] = 162.23         # add a new key
drug_mw["aspirin"] = 180.2           # change an existing value
print(drug_mw)

print("ibuprofen" in drug_mw)        # checks the keys
print(len(drug_mw))
````

### Keys, values and items

````python
print(list(drug_mw.keys()))
print(list(drug_mw.values()))
print(list(drug_mw.items()))
````

### Nested dictionaries

A value can be another dictionary. This is a good way to store several properties for each molecule.

````python
molecules = {
    "aspirin":     {"smiles": "CC(=O)Oc1ccccc1C(=O)O",       "mw": 180.16},
    "caffeine":    {"smiles": "Cn1cnc2c1c(=O)n(C)c(=O)n2C",  "mw": 194.19},
    "paracetamol": {"smiles": "CC(=O)Nc1ccc(O)cc1",          "mw": 151.16},
}

print(molecules["caffeine"]["smiles"])
````

## Part 5: Loops

A **loop** repeats code for each item in a collection, so you do not have to write the same line many times.

```python
for item in collection:
    # code here runs once for each item
```

The lines inside the loop must be **indented** (4 spaces, Colab does this for you after the `:`).

````python
drugs = ["aspirin", "caffeine", "ibuprofen", "paracetamol"]

for drug in drugs:
    print("Testing", drug)

print("Done")     # not indented, so it runs once after the loop
````

### Looping over a string

A loop over a string gives you one character at a time.

````python
for base in "ATGC":
    print(base)
````

### `range()`

`range(n)` gives the numbers 0, 1, ..., n-1. `range(start, stop, step)` lets you jump. This is useful for reading a sequence **codon by codon**.

````python
for i in range(3):
    print(i)
````

````python
# Print only the first base of each codon of the following DNA sequence

dna = "ATGGCGTACGCTTAGCCGATATAA"
for i in range(0, len(dna), 3):
    print(dna[i])
````

````python
# Print each 3-base codon from the DNA sequence
for i in range(0, len(dna), 3):
    # Print the full 3-base codon
    print(dna[i:i+3])
````

### Looping over a dictionary

Use `.items()` to get the key and the value together.

````python
drug_mw = {"aspirin": 180.16, "caffeine": 194.19, "ibuprofen": 206.28, "paracetamol": 151.16}

for name, mw in drug_mw.items():
    print(f"{name}: {mw} g/mol")
````

### The accumulator pattern

Start with an empty value before the loop, then add to it on each step. This is one of the most common patterns in data analysis.

````python
total = 0
for mw in drug_mw.values():
    total += mw
print("Total MW:", total)
````

````python
rna = ""
dna = "ATGGCGTACGCTTAGCCGATATAA"
for base in dna:
    if base == "T":
        rna += "U"
    else:
        rna += base
print(rna)
````

### Making decisions with `if`

The last example used `if`. It runs code only when a condition is `True`.

```python
if condition:
    ...
elif other_condition:
    ...
else:
    ...
```

| Comparison | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` `>` | less than, greater than |
| `<=` `>=` | less or equal, greater or equal |

Combine conditions with `and`, `or`, `not`.

````python
for name, mw in drug_mw.items():
    if mw < 160:
        size = "fragment"
    else:
        size = "not a fragment"
    print(f"{name} ({mw}) is {size}")
````

### Building a list inside a loop

Start with an empty list and `append()` items that pass a test.

````python
light_drugs = []
for name, mw in drug_mw.items():
    if mw < 190:
        light_drugs.append(name)
print(light_drugs)
````

### `while` loops (short note)

A `while` loop repeats as long as a condition is `True`. We use `for` loops much more often, but `while` is useful when you do not know how many steps you need.

````python
user_input = ""

# Loops endlessly until the user explicitly types 'exit'
while user_input.lower() != "exit":
    user_input = input("Type something (or 'exit' to quit): ")
    print(f"You typed: {user_input}")
````
